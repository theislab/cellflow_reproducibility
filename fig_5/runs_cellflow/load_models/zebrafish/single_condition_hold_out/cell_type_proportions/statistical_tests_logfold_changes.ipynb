```python
unique_cts = df["cell_type_broad"].unique()
```

```python
df.index = df["cell_type_broad"]
```

```python
cf_stats = df[df["method"]=="cellflow"].loc[unique_cts]["pearson"].values
cl_stats = df[df["method"]=="cl"].loc[unique_cts]["pearson"].values
```

```python
ttest_rel(cf_stats, cl_stats)
```

```python
import numpy as np
mask = ~np.isnan(cf_stats) & ~np.isnan(cl_stats)
a_clean = cf_stats[mask]
b_clean = cl_stats[mask]
```
