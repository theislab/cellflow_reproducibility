```python
import pandas as pd
from scipy.stats import ttest_rel

```

```python
df = pd.read_csv("/lustre/groups/ml01/workspace/ot_perturbation/data/zebrafish_new/results/allocation/all_models_metrics.csv", index_col=0)
```

```python
df.head()
```

```python
df["cell_type_accuracy"] = (2.0 - df["cell_type_fraction_error"])/2.0
df = df[["cell_type_accuracy", "model"]]
df["condition"] = df.index
```

```python
unique_conditions = df["condition"].unique()
```

```python
df["model"].unique()
```

```python
unique_conditions = list(df.index.unique())
```

```python
len(unique_conditions)
```

```python
cf_stats = df[df["model"]=="cf"].loc[unique_conditions]["cell_type_accuracy"].values
id_stats = df[df["model"]=="id"].loc[unique_conditions]["cell_type_accuracy"].values
mean_tp_stats = df[df["model"]=="same_timepoint"].loc[unique_conditions]["cell_type_accuracy"].values
mean_p_stats = df[df["model"]=="same_perturbation"].loc[unique_conditions]["cell_type_accuracy"].values
```

```python
ttest_rel(cf_stats, id_stats)
```

```python
ttest_rel(cf_stats, mean_tp_stats)
```

```python
ttest_rel(cf_stats, mean_p_stats)
```
