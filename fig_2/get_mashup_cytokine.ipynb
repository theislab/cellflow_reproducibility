```python
import pickle
import scanpy as sc
```

```python
cytokine_held_out = "IFN-beta"
adata_base = sc.read_h5ad(f"/lustre/groups/ml01/workspace/ot_perturbation/data/pbmc/new_cytokine/adata_base_{cytokine_held_out}.h5ad")
adata_rest = sc.read_h5ad(f"/lustre/groups/ml01/workspace/ot_perturbation/data/pbmc/new_cytokine/adata_rest_{cytokine_held_out}.h5ad")
cytokines = set(adata_base.obs["cytokine"].unique()).union(set(adata_rest.obs["cytokine"].unique()))
```

```python
cytokines = set(adata_base.obs["cytokine"].unique()).union(set(adata_rest.obs["cytokine"].unique()))
```

```python
with open("/lustre/groups/ml01/workspace/ot_perturbation/data/embeddings/Mashup_embeddings.pkl", 'rb') as f:
    mashup_embeddings = pickle.load(f)

with open("/lustre/groups/ml01/workspace/ot_perturbation/data/embeddings/dict_cytokines.pickle", 'rb') as f:
    dict_cytokines = pickle.load(f)
```

```python
cyto_renamings = {}
```

```python
cytos_not_found = []
for cyto in mashup_embeddings.keys():
    if cyto in cytokines:
        cyto_renamings[cyto] = cyto
    elif "IL" in cyto:
        cyto_adapted = "IL-"+cyto.strip("IL")
        if cyto_adapted in cytokines:
            cyto_renamings[cyto] = cyto_adapted
        else:
            cytos_not_found.append(cyto)
    elif cyto in dict_cytokines.keys():
        if dict_cytokines[cyto] in cytokines:
            cyto_renamings[cyto] = dict_cytokines[cyto]
            assert dict_cytokines[cyto] in cytokines
        else:
            cytos_not_found.append(cyto)
       
    else:
        cytos_not_found.append(cyto)
```
