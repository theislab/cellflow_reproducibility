```python
import pandas 
import pickle
import pandas as pd
import os
path = "/lustre/groups/ml01/workspace/ot_perturbation/data/embeddings/cache/pathway_embeddings"
```

```python
gene_embs = ["c7.immunesigdb.v2023.2.Hs.symbols.pkl",
#"c3.mir.mirdb.v2023.2.Hs.symbols.pkl",
"c7.all.v2023.2.Hs.symbols.pkl",
"c3.all.v2023.2.Hs.symbols.pkl", # (this has all)
"c5.go.v2023.2.Hs.symbols.pkl", #(this has all)
"c1.all.v2023.2.Hs.symbols.pkl", # (this has all)
"c5.go.bp.v2023.2.Hs.symbols.pkl", 
"c3.tft.v2023.2.Hs.symbols.pkl", # (this has all)
"c3.tft.gtrd.v2023.2.Hs.symbols.pkl",
"c2.cgp.v2023.2.Hs.symbols.pkl",
"stringdb.human.medium.pkl"]
```

```python
alias_dict = {"FOXL2NB": ["C3orf72", "C3ORF72", "FLJ43329", "FOXL2"],
             "CBARP": ["BARP", "DOS", "C19orf26", "C19ORF26"],
             "SAMD1": [],
             "CITED1": ["MSG1"],
             "CLDN6": ["Claudin 6", "Claudin-6"],
             "CEBPE": ["CRP1"],
             "MIDEAS": ["C14orf117", "C14orf43", "ELMSAN1", "LSR68"],
             }
```

```python
for emb in gene_embs:
    df = pd.read_pickle(os.path.join(path, emb))
    diff = set(ensembl_to_gene.values()) - set(df.index)
    for el in diff:
        if el == "ctrl":
            continue
        aliases = alias_dict[el]
        print("aliases found: ", set(ensembl_to_gene.values()).intersection(set(aliases)))
    print(emb, diff-set(("ctrl", )))
```

```python
aliases
```

```python
dfs = {}
genes = list(set(ensembl_to_gene.values()) - set(("ctrl", )))
for emb in gene_embs:
    df = pd.read_pickle(os.path.join(path, emb))
    diff = set(genes) - set(df.index)
    df_red = df[df.index.isin(ensembl_to_gene.values())]
    dfs_to_append = []
    for el in diff:
        aliases = alias_dict[el]
        aliases = list(set(aliases).intersection(set(genes)))
        if len(aliases) > 1:
            print(el, aliases)
            raise ValueError
        if len(aliases) == 1:
            dff = df.loc[[aliases[0]]]
            dff.index = [el]
            dfs_to_append.append(dff)
        else:
            df_tmp = pd.DataFrame(df_red.mean(axis=0)).T
            df_tmp.index = [el]
            dfs_to_append.append(df_tmp)

    if len(dfs_to_append):
        df_all = pd.concat((df_red, pd.concat(dfs_to_append, axis=0)), axis=0)
        df_all2 = df_all.copy()
    else:
        df_all = df_red
    dfs[emb] = df_all.loc[genes]
        
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

# Step 1: Get correlation matrices for all datasets
corrs = {name: df.corr() for name, df in dfs.items()}

# Step 2: Reference correlation matrix and genes
ref_name, ref_corr = next(iter(corrs.items()))
genes = ref_corr.index[:10]  # First 10 genes

# Step 3: Compute similarities for each gene across datasets
fig, axes = plt.subplots(5, 2, figsize=(14, 20))
axes = axes.flatten()

for i, gene in enumerate(genes):
    similarities = []
    dataset_names = []
    
    ref_vec = ref_corr.loc[gene].values.reshape(1, -1)  # shape (1, 102)
    
    for name, corr in corrs.items():
        if gene in corr.index:
            vec = corr.loc[gene].values.reshape(1, -1)
            sim = cosine_similarity(ref_vec, vec)[0, 0]
            similarities.append(sim)
            dataset_names.append(name)
    
    ax = axes[i]
    sns.barplot(x=similarities, y=dataset_names, ax=ax)
    ax.set_title(f"Similarity of {gene} across datasets")
    ax.set_xlim(0, 1)  # cosine similarity in [0, 1]
    ax.set_xlabel("Cosine similarity")
    ax.set_ylabel("Dataset")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from scipy.cluster.hierarchy import linkage, leaves_list

# Get the ordering from the first DataFrame
first_df = list(dfs.values())[8]
corr_first = first_df.corr()

# Compute linkage and order
linkage_rows = linkage(corr_first, method='average')
ordered_indices = leaves_list(linkage_rows)
ordered_labels = corr_first.index[ordered_indices]

# Create a 3x3 subplot grid
fig, axes = plt.subplots(3, 3, figsize=(15, 12))
axes = axes.flatten()

# Plot each heatmap using the same order
for i, (name, df) in enumerate(dfs.items()):
    if i >= 9:
        break
    corr = df.corr()
    # Reorder rows and columns to match first_df clustering
    corr_ordered = corr.loc[ordered_labels, ordered_labels]
    sns.heatmap(corr_ordered, ax=axes[i], cmap='coolwarm', cbar=False)
    axes[i].set_title(name)

# Remove unused axes if fewer than 9
for j in range(len(dfs), 9):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```
